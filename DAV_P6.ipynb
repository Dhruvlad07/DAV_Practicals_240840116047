#Practical no :- 6 | Create basic data visualizations (Line, Bar, Histogram, Scatter, Pie).

Step 1: import Required Libries

```python
# Import the pandas, numpy, matplotlib.pyplot library for data manipulation
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Step 2: Create a Dataset

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun','Jul','Aug','Sep','Oct','Nov','Dec'],
    'Product': ['Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet'],
    'Region': ['North','South','East','West','North','South','East','West','North','South','East','West'],
    'Sales': [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    'Unit_Sold': [12,30,25,14,35,30,18,38,32,21,45,40]
}
# Create a pandas DataFrame from the dictionary
df = pd.DataFrame(data)
df # Display the DataFrame
```

Tast 2 Create a line chart.

2.1 How did the company sales change from Jan to Dec.



```python
#Creating a line chart for Company Sales.
plt.figure(figsize=(8, 4))
plt.plot(df['Month'], df['Sales'], marker='o', linestyle='-', color='b')
plt.title('Company Sales from Jan to Dec')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Task 3 Create a Bar Chart.

3.1 Which product generates how much total sales (Groupby)

```python
product_sales = df.groupby('Product')['Sales'].sum().reset_index() # Group by 'Product' and sum 'Sales'
plt.figure(figsize=(8, 4)) # Set the figure size
plt.bar(product_sales['Product'], product_sales['Sales'], color=['skyblue', 'lightcoral', 'lightgreen']) # Create a bar chart of total sales by product
plt.title('Total Sales by Product') # Set the title of the chart
plt.xlabel('Product') # Set the x-axis label
plt.ylabel('Total Sales') # Set the y-axis label
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add a horizontal grid
plt.tight_layout() # Adjust layout
plt.show() # Display the plot
```

Task 4 Create a histogram.

4.1 How are the individual sales values distributed

```python
plt.figure(figsize=(8, 4)) # Set the figure size
plt.hist(df['Sales'], bins=5, color='Yellow', edgecolor='black') # Create a histogram of sales values with 5 bins
plt.title('Distribution of Sales Values') # Set the title of the histogram
plt.xlabel('Sales') # Set the x-axis label
plt.ylabel('Frequency') # Set the y-axis label
plt.grid(axis='y', alpha=0.75) # Add a horizontal grid
plt.tight_layout() # Adjust layout
plt.show() # Display the plot
```

Task 5 Create a Scater plot.

5.1 Is there are relationship between the number of units sold and total sales.

```python
plt.figure(figsize=(8, 4)) # Set the figure size
plt.scatter(df['Unit_Sold'], df['Sales'], color='green', alpha=0.7) # Create a scatter plot of 'Unit_Sold' vs 'Sales'
plt.title('Relationship between Units Sold and Total Sales') # Set the title of the scatter plot
plt.xlabel('Units Sold') # Set the x-axis label
plt.ylabel('Total Sales') # Set the y-axis label
plt.grid(True) # Add a grid
plt.tight_layout() # Adjust layout
plt.show() # Display the plot
```

Task 6 Create a PIE Chart.

6.1 What % of the total sales comes from each region.

```python
region_sales = df.groupby('Region')['Sales'].sum() # Group by 'Region' and sum 'Sales'
plt.figure(figsize=(4, 4)) # Set the figure size for the pie chart
plt.pie(region_sales, labels=region_sales.index, autopct='%1.1f%%', startangle=90, colors=['gold', 'lightcoral', 'lightskyblue', 'lightgreen']) # Create a pie chart of sales by region
plt.title('Percentage of Total Sales by Region') # Set the title of the pie chart
plt.axis('equal')  # Ensure the pie chart is circular.
plt.tight_layout() # Adjust layout
plt.show() # Display the plot
```
