Step 1 : Import Pandas

Step 2 : Create Dataset/Records

```python
#Creating a DataFrame
import pandas as pd

data = {
    "ID": [101,102,103,104,104,105,106,107],
    "Name": ["Rahul","Priya","Amit",None,"Riya","Ajay","Naresh","Jenil"],
    "Age": [20,21,None,20,20,22,23,28],
    "City": ["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahmedabad"],
    "Marks":[85,None,78,88,88,81,90,93]
}
df=pd.DataFrame(data)
print(df)
```

Step 3 : Inspect the Dataset

```python
#Display First five rows of the dataset
df.head()
```

Step 4 : Check Missing Values

```python
#Check missing values for all the columns
df.isnull()
```

```python
#Check missing values for each columns
df.isnull().sum()
```

```python
#Print the total number of count of missing values
df.isnull().sum().sum()
```

Step 5 : Fill Missing Values

```python
#Fill missing values
# df.fillna(0)
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df
```

Fill Missing Age

```python
#Replace missing value in the Age

df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

Fill Missing Name

```python
# Replace missing name with the text Unknown

df = df.fillna("Unknown")
df
```

Step 6 : Detect duplicate rows

```python
#Find duplicate values
df.duplicated()
```

```python
#Count the total number of duplicate rows
df.duplicated().sum()
```

Step 7 : Remove Duplicate Records

```python
#Remove the duplicate records
df.drop_duplicates()
```

Step 8 : Rename a column

```python
#Rename the column marks to sscore
df.rename(columns={"Marks":"Score"}, inplace= True)
df
```

Step 9 : Change Data Type

```python
# Convert the Age column from float to integer
# This removes the decimal part after filling missing values
df['Score'] = df['Score'].astype(int)
df.dtypes
```

Step 10 : Add final score column by adding 5 marks

```python
# Create a new column
# Add 5 marks to the column "Score"
df['Final_Score']= df['Score'] + 5
df
```

Step 11 : Convert text to uppercase

```python
# Conversion of name from given case to uppercase
df['Name'] = df['Name'].str.upper()
df
```

Step 12 : Convert text to lowercase

```python
# Conversion of city from given case to lowercase
df['City'] = df['City'].str.lower()
df
```

Step 13 : Remove Extra Space

```python
# Removing extra space
df['City'] = df['City'].str.strip()
df
```

Step 14 : Sort the dataset

```python
# sorting the data based on "Final Score"
df.sort_values(by="Final_Score", ascending=False)
```

Step 15 : Save the cleaned dataset

```python
# Save cleaned dataset to CSV file
df.to_csv("clean_student_data.csv", index=False)
print("Dataset saved successfully")
```
