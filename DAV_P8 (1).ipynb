Step: 1 Importing Required Libraries

```python
# Import the pandas, numpy, matplotlib.pyplot library for data manipulation
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
import plotly.express as px
```

Step: 2 Creating Dataset

```python
data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Product': ['Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet','Laptop','Mobile','Tablet'],
    'Region': ['North','South','East','West','North','South','East','West','North','South','East','West'],
    'Sales':[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    'Unit_Sold':[12,30,25,14,35,30,18,38,32,21,45,40],
    'Profit':[18000,27000,30000,22000,32000,35000,28000,34000,38000,33000,46000,50000]
}
df = pd.DataFrame(data)
df
```

```python
#Display the first five record
df.head()
```

Task 3 Create an Interactive Line Chart using Ploty Express

```python
#How does sales revenue change throughout the year?
fig = px.line(df, x='Month', y='Sales', title='Sales Revenue Over Time')
fig.show()
```

Task 4 Improve the Hover information

```python
# Create a line chart using sales over month.
fig = px.line(df, x='Month', y='Sales', title='Sales Revenue Over Time')
fig.update_traces(
    hovertemplate='Month: %{x}<br>Sales: %{y}'
)
fig.show()
```

```python
#add additional columns to the hover information. Student can see Product, Region,Units sold and profit when they move the mouse over a data point.
fig = px.line(df, x='Month', y='Sales', title='Sales Revenue Over Time', hover_data=['Product', 'Region', 'Unit_Sold', 'Profit'])
fig.show()
```

Task 5 Create an Interactive Bar Chart

```python
# Group the dataset according to Product.
product_data = df.groupby('Product')[['Sales', 'Unit_Sold', 'Profit']].sum().reset_index()

# Create an interactive bar chart of Sales by Product
fig = px.bar(product_data, x='Product', y='Sales',
             title='Total Sales Revenue by Product',
             labels={'Sales': 'Total Sales ($)', 'Product': 'Product Type'},
             text_auto='.2s',
             color='Product')

fig.show()
product_data
```

```python
#Calculate the total sales generated by each product
total_sales = df.groupby('Product')['Sales'].sum()
total_sales
```

```python
#Create an interacive bar chart.
fig = px.bar(total_sales, x=total_sales.index, y=total_sales.values,
             title='Total Sales Revenue by Product',
             labels={'x': 'Product Type', 'y': 'Total Sales ($)'})
fig.show()
```

Task 6 Interactive Scatter Plot

```python
#To analyze the relationship between units sold and sales using an interactive scatter plot.
fig = px.scatter(df, x='Unit_Sold', y='Sales',
                  title='Relationship between Units Sold and Sales',
                  labels={'Unit_Sold': 'Units Sold', 'Sales': 'Sales ($)'})
fig.show()
```
