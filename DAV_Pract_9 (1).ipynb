**Step 2: Import libraries**

```python
# Import Pandas to handle data
import pandas as pd
# Import Plotly Express to create graphs
import plotly.express as px
# Import Dash components to create the dashboard
from dash import Dash, dcc, html, Input, Output
```

**Step 3: Create the business dataset**

```python
# Create a sample business dataset
data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile",
                "Tablet", "Laptop", "Mobile", "Tablet", "Laptop"],

    "Region": ["North", "South", "East", "West", "North",
               "South", "East", "West", "South", "North"],

    "Sales": [50000, 30000, 20000, 45000, 35000,
              25000, 60000, 40000, 22000, 55000],

    "Profit": [10000, 6000, 4000, 9000, 7000,
               5000, 12000, 8000, 4400, 11000]
}

# Convert the dataset into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

**Step 4: Create the dashboard layout**

```python
# Create a Dash application
app = Dash(__name__)

# Design the dashboard layout
app.layout = html.Div([

    # Add the dashboard title
    html.H1(
        "Business Reporting Dashboard",
        style={"textAlign": "center", "color": "blue"}
    ),

    # Add a label for the region filter
    html.Label("Select Region:"),

    # Create a dropdown to select a region
    dcc.Dropdown(
        id="region",
        options=[
            {"label": "All Regions", "value": "All"}
        ] + [
            {"label": r, "value": r}
            for r in sorted(df["Region"].unique())
        ],
        value="All",
        clearable=False
    ),

    # Display sales, profit, and record count
    html.Div(id="summary"),

    # Add a graph for sales by product
    dcc.Graph(id="sales_chart"),

    # Add a graph for profit distribution
    dcc.Graph(id="profit_chart"),

    # Add a graph for profit by product
    dcc.Graph(id="product_chart")

], style={"padding": "20px"})
```

**Step 5: Add interactive dashboard functionality**

```python
# Update the dashboard whenever the region changes
@app.callback(
    Output("summary", "children"),
    Output("sales_chart", "figure"),
    Output("profit_chart", "figure"),
    Output("product_chart", "figure"),
    Input("region", "value")
)

# Define the function to update dashboard data and charts
def update_dashboard(region):

    # Filter the dataset according to the selected region
    if region == "All":
        filtered = df
    else:
        filtered = df[df["Region"] == region]

    # Calculate total sales and profit
    total_sales = filtered["Sales"].sum()
    total_profit = filtered["Profit"].sum()

    # Display the summary information
    summary = html.Div([
        html.H3(f"Total Sales: ₹{total_sales:,}"),
        html.H3(f"Total Profit: ₹{total_profit:,}"),
        html.H3(f"Total Records: {len(filtered)}")
    ])

    # Create a bar chart for sales by product
    sales_fig = px.bar(
        filtered,
        x="Product",
        y="Sales",
        color="Region",
        title="Sales by Product"
    )

    # Create a pie chart for profit distribution by region
    profit_fig = px.pie(
        filtered,
        names="Region",
        values="Profit",
        title="Profit Distribution by Region"
    )

    # Calculate total profit for each product
    product_data = filtered.groupby(
        "Product", as_index=False
    )["Profit"].sum()

    # Create a bar chart for profit by product
    product_fig = px.bar(
        product_data,
        x="Product",
        y="Profit",
        color="Product",
        title="Profit by Product"
    )

    # Return the summary and all three charts
    return summary, sales_fig, profit_fig, product_fig
```

**Step 6: Run the dashboard in Google Colab**

```python
# Run the dashboard and display it inside Google Colab
app.run(jupyter_mode="inline", debug=False, port=8050)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Set plotting style
sns.set_theme(style="whitegrid")

# Create a master figure that replicates the dashboard layout
fig = plt.figure(figsize=(15, 12), dpi=300)
fig.suptitle("Business Reporting Dashboard (Step 6 Export)", fontsize=24, color="blue", weight="bold")

# Add KPI text at the top
kpi_text = "Total Sales: ₹382,000      |      Total Profit: ₹76,800      |      Total Records: 10"
fig.text(0.5, 0.92, kpi_text, ha="center", va="center", fontsize=14, weight="bold",
         bbox=dict(boxstyle="round,pad=0.5", facecolor="#f0f8ff", edgecolor="blue", lw=1))

# 1. Sales by Product (Bar Chart)
ax1 = plt.subplot2grid((2, 2), (0, 0))
sales_data = df.groupby(["Product", "Region"])["Sales"].sum().reset_index()
sns.barplot(data=sales_data, x="Product", y="Sales", hue="Region", ax=ax1)
ax1.set_title("Sales by Product", fontsize=14, weight="bold")
ax1.set_xlabel("Product", fontsize=12)
ax1.set_ylabel("Sales (₹)", fontsize=12)
ax1.legend(title="Region")

# 2. Profit Distribution by Region (Pie Chart)
ax2 = plt.subplot2grid((2, 2), (0, 1))
region_profit = df.groupby("Region")["Profit"].sum()
ax2.pie(region_profit, labels=region_profit.index, autopct='%1.1f%%',
        colors=sns.color_palette("pastel"), startangle=90, textprops={'fontsize': 12, 'weight': 'bold'})
ax2.set_title("Profit Distribution by Region", fontsize=14, weight="bold")

# 3. Profit by Product (Bar Chart)
ax3 = plt.subplot2grid((2, 2), (1, 0), colspan=2)
product_profit = df.groupby("Product")["Profit"].sum().reset_index()
sns.barplot(data=product_profit, x="Product", y="Profit", palette="viridis", ax=ax3)
ax3.set_title("Profit by Product", fontsize=14, weight="bold")
ax3.set_xlabel("Product", fontsize=12)
ax3.set_ylabel("Profit (₹)", fontsize=12)

# Adjust layout and save the master dashboard image
plt.tight_layout(rect=[0, 0, 1, 0.90])
plt.savefig("complete_dashboard.png", bbox_inches="tight", dpi=300)
plt.close()

# Generate a clean, complete, and responsive HTML version of the dashboard
html_content = """
<!DOCTYPE html>
<html>
<head>
    <title>Business Reporting Dashboard</title>
    <script src="https://cdn.plot.ly/plotly-latest.min.js"></script>
    <style>
        body { font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; margin: 0; padding: 20px; background-color: #f4f6f9; }
        .header { text-align: center; color: blue; margin-bottom: 25px; }
        .kpi-container { display: flex; justify-content: space-around; margin-bottom: 30px; }
        .kpi-card { background: white; padding: 20px; border-radius: 8px; box-shadow: 0 4px 6px rgba(0,0,0,0.1); text-align: center; width: 28%; border-top: 5px solid blue; }
        .kpi-card h3 { margin: 0; color: #555; font-size: 16px; }
        .kpi-card p { margin: 10px 0 0 0; font-size: 24px; font-weight: bold; color: #222; }
        .chart-row { display: flex; justify-content: space-between; margin-bottom: 20px; }
        .chart-container { background: white; padding: 15px; border-radius: 8px; box-shadow: 0 4px 6px rgba(0,0,0,0.1); width: 48%; }
        .full-chart { background: white; padding: 15px; border-radius: 8px; box-shadow: 0 4px 6px rgba(0,0,0,0.1); width: 100%; }
    </style>
</head>
<body>
    <div class="header">
        <h1>Business Reporting Dashboard</h1>
    </div>
    <div class="kpi-container">
        <div class="kpi-card"><h3>Total Sales</h3><p>₹3,82,000</p></div>
        <div class="kpi-card"><h3>Total Profit</h3><p>₹76,800</p></div>
        <div class="kpi-card"><h3>Total Records</h3><p>10</p></div>
    </div>
    <div class="chart-row">
        <div id="sales_chart_div" class="chart-container"></div>
        <div id="profit_chart_div" class="chart-container"></div>
    </div>
    <div id="product_chart_div" class="full-chart"></div>

    <script>
"""

# Create Plotly configuration scripts to load in HTML
import json
import plotly.utils as utils

# Convert plotly objects to JSON compatible with javascript representation
sales_json = json.dumps(sales_fig, cls=utils.PlotlyJSONEncoder)
profit_json = json.dumps(profit_fig, cls=utils.PlotlyJSONEncoder)
product_json = json.dumps(product_fig, cls=utils.PlotlyJSONEncoder)

html_content += f"""
        var sales_fig = {sales_json};
        var profit_fig = {profit_json};
        var product_fig = {product_json};

        Plotly.newPlot('sales_chart_div', sales_fig.data, sales_fig.layout, {{responsive: true}});
        Plotly.newPlot('profit_chart_div', profit_fig.data, profit_fig.layout, {{responsive: true}});
        Plotly.newPlot('product_chart_div', product_fig.data, product_fig.layout, {{responsive: true}});
    </script>
</body>
</html>
"""

with open("dashboard_report.html", "w", encoding="utf-8") as f:
    f.write(html_content)

print("\n✅ SUCCESS!")
print("- Saved complete static dashboard as: complete_dashboard.png")
print("- Saved complete interactive web dashboard template as: dashboard_report.html")
print("\nPlease open the file explorer in the left sidebar to download these files!")
```
