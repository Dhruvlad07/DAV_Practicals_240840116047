Step 1 : Import Pandas

```python
# Import the Pandas library under the alias 'pd' for data manipulation and analysis
import pandas as pd
```

```python
# Define raw student dataset structured as a dictionary of lists (key=column name, value=row values)
data = {
    "ID": [101,102,103,104,105,106,107],
    "Name": ["Rahul", "Priya", "Amit", "Riya", "Karan", "Krisha", "Harry"],
    "Age": [20,21,19,20,22,24,28],
    "City": ["Surat", "Navsari", "Valsad", "Bharuch", "Surat", "Gandhinagar", "Ahmedabad"],
    "Marks": [85,92,78,88,81,84,90]
}
```

Step 3: Create a DataFrame

```python
# Convert the dictionary into a 2D tabular Pandas DataFrame structure and render it
df = pd.DataFrame(data)

df #or you can type print(df)
```

Step 4 : Save Data as CSV File

```python
# Export DataFrame to a CSV file named 'student.csv' without including the default integer index
df.to_csv("student.csv", index=False)
print("CSV file created successfully")
```

Step 5 : Save Data as Excel File

```python
# Export DataFrame to an Excel file named 'student.xlsx' without including the index
df.to_excel("student.xlsx", index=False)
print("Excel file created successfully")
```

Step 6 : Save Data as JSON File

```python
# Export DataFrame to a JSON file named 'student.json' using a records orientation with indentation
df.to_json("student.json", orient="records", index=4)
print("JSON file created successfully")
```
