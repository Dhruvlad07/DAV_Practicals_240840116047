#Practical No 3 Apply Filtering, Sorting, Aggregation, and GroupBy Operation of real world data set

Step 1: Create a Dataframe/Records

```python
#Create Data Set
import pandas as pd
import numpy as np
data = {
    "Emp_ID": [101,102,103,104,105,106,107,108],
    "Name": ["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department": ["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City": ["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary": [45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
df = pd.DataFrame(data)
#Diplay Data Set
df
```

Step 2: Filter Data

2.1 Display employee from IT Department

```python
#Select only employeea working in the It Depatrment
#The condition returns rows where Department is equal to "IT"

df[df["Department"]=="IT"]
```

2.2 Display Employees with salary greater than 50000

```python
#Display employee whose salary is greater than 50000

df[df["Salary"]>50000]
```

2.3 Display Employee from Surat

```python
#Display Employee whose city is surat
df[df["City"]=="Surat"]
```

2.4 Apply Multiple Condition

```python
#Display employee fropm IT Department
#Having Salary Greater then 50000
df[ (df["Department"] == "IT") & (df["Salary"]>50000)]
```

Step 3: Sorting Data

3.1 Sort Salary in Ascending Order

```python
#Sort employee salary from lowest to highest
df.sort_values(by='Salary')
```

3.2 Sort Salary in descending order

```python
#Sort salary to Descending Order
df.sort_values(by='Salary', ascending=False)
```

3.3 Sort Department Wise Salary

```python
df.sort_values(by=['Department','Salary'], ascending=[True,False])
```

Step 4 Perfrom Aggregarion

4.1 Calculate Total Salary

```python
#Calculate total salary

df["Salary"].sum()
```

4.2 Calculate the avg salary

```python
# Calculate the average salary by mean
df["Salary"].mean()
```

4.3 Find Highest Salary

```python
# Find the highest salary by max
df["Salary"].max()
```

4.4 Find Lowest Salary

```python
# Find the lowest salary by min
df["Salary"].min()
```

4.5 Count Employees

```python
#Count Empoloyees
df["Salary"].count()
```

Step 5 Perfrom Group By Operation

5.1 Group Employee Based on department and Claculate the avg salary of each department

```python
#Group employees based on Department
#Calculate the average salary of each department
df.groupby("Department")["Salary"].mean()
```

5.2 Calculate the total salary paid by department

```python
#Calculate the total salary paid by each department
df.groupby("Department")["Salary"].sum()
```

5.3 Display the highest salary in each department

```python
#Display the highest salary in each department
df.groupby("Department")["Salary"].max()
```

5.4 Count Employees Department Wise

```python
#count the number of employees in each department.
df.groupby("Department")["Emp_ID"].count()
```

5.5 Calculate the average salary of employees in each city

```python
#Calculate the average salary of employee in each city.
df.groupby("City")["Salary"].mean()
```

Step 6 Multpile Aggregartion Functions

```python
df.groupby("Department")["Salary"].agg(['count','sum','min','max','mean'])
```

Step 7 Save the Result

```python
#Save the dataset into csv file
df.to_csv("Employee_Data.csv",index=False)
print("Dataset saved successfully.")
```
